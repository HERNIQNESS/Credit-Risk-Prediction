## DATA CLEANING
- Date columns to datetime
- splitting indicators to better see all affecting indicators in the dataset
- pivoting and merging to show relationships
- saving the cleaned dataset

```python
# LIB N DATA

import pandas as pd
import numpy as np

train = pd.read_csv(r'C:\Users\User\Desktop\credit-risk-project\data\raw\train.csv')
economic = pd.read_csv(r'C:\Users\User\Desktop\credit-risk-project\data\raw\economic_indicators.csv')

```

```python
train.head()
```

```python
# DATA TYPES N MISSING VALUES IN TRAIN
print("Data types:\n", train.dtypes)
print("\nMissing values:\n", train.isnull().sum())
```

```python
# DATE FORMATTING STR TO DATETIME
train['disbursement_date'] = pd.to_datetime(train['disbursement_date'])
train['due_date'] = pd.to_datetime(train['due_date'])
print(train[['disbursement_date', 'due_date']].dtypes)
print(train[['disbursement_date', 'due_date']].head())
```

```python
print("Data types:\n", train.dtypes)
```

```python
# DUPLICATE RECS
print("Duplicate rows:", train.duplicated().sum())
print("Duplicate loan IDs:", train['tbl_loan_id'].duplicated().sum())
print("Duplicate customer IDs:", train['customer_id'].duplicated().sum())
```

```python
economic.head()
```

```python
# RESHAPING ECON DATASET, (COLUMNS R YEARS)
economic_melted = economic.melt(
    id_vars=['Country', 'Indicator'],
    var_name='Year',
    value_name='Value'
)
economic_melted['Year'] = economic_melted['Year'].str.replace('YR', '').astype(int)
print(economic_melted.shape)
economic_melted.head(10)
```

```python
# PIVOT ECONOMIC DATASET SO EACH INDICATOR IS A COLUMN(i.e: inflation, consumer prices, Real interest rate, unemployment rate etc)
economic_pivot = economic_melted.pivot_table(
    index=['Country', 'Year'],
    columns='Indicator',
    values='Value'
).reset_index()
economic_pivot.columns.name = None
print(economic_pivot.shape)
economic_pivot.head()
```

```python
# Extract year from train to merge on
train['loan_year'] = train['disbursement_date'].dt.year
print(train['loan_year'].value_counts().sort_index())
```

```python
# PROCESSED DATA
train.to_csv(r'C:\Users\User\Desktop\credit-risk-project\data\processed\train_cleaned.csv', index=False)
economic_pivot.to_csv(r'C:\Users\User\Desktop\credit-risk-project\data\processed\economic_pivot.csv', index=False)
print("Saved successfully")
```
