## EXPLORATORY DATA ANALYSIS
- exploring the structure and patterns in the dataset.
- Analysis plotting distribution of targets

```python
# LIB N DATA
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

train = pd.read_csv(r'C:\Users\User\Desktop\credit-risk-project\data\processed\train_cleaned.csv')
```

```python
# TARGET DISTRIBUTION - IMBALANCE...... 0'S VS 1'S DEFAULTED
plt.figure(figsize=(6,4))
sns.countplot(x='target', data=train)
plt.title('Target Distribution')
plt.xticks([0,1], ['No Default', 'Default'])
plt.show()
print(train['target'].value_counts())
```

```python
# LOAN AMOUNT N DEFAULT
plt.figure(figsize=(8,4))
sns.boxplot(x='target', data=train, y='Total_Amount')
plt.title('Loan Amount vs Default')
plt.xticks([0,1], ['No Default', 'Default'])
plt.show()
```

```python
# LOAN TYPE AND DEFAULT
plt.figure(figsize=(8,4))
ct = pd.crosstab(train['loan_type'], train['target'], normalize='index') * 100
ct.plot(kind='bar', stacked=True)
plt.title('Default Rate by Loan Type')
plt.ylabel('% percentage')
plt.legend(['No Default', 'Default'])
plt.tight_layout()
plt.show()
```

```python
# OLD VS NEW VS REPEAT BORROWERS
plt.figure(figsize=(6,4))
ct2 = pd.crosstab(train['New_versus_Repeat'], train['target'], normalize='index') * 100
ct2.plot(kind='bar', stacked=True)
plt.title('Default Rate: New vs Repeat Borrowers')
plt.ylabel('% percentage')
plt.legend(['No Default', 'Default'])
plt.tight_layout()
plt.show()
```

```python
# DURATION AND DEFAULTING 
plt.figure(figsize=(8,4))
sns.boxplot(x='target', data=train, y='duration')
plt.title('Loan Duration vs Default')
plt.xticks([0,1], ['No Default', 'Default'])
plt.show()
```

```python
# CORRELATION
plt.figure(figsize=(10,6))
numeric_cols = train.select_dtypes(include=np.number).columns.tolist()
sns.heatmap(train[numeric_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```
