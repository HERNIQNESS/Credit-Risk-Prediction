## FEATURE ENGINEERING

```python
import pandas as pd
import numpy as np

train = pd.read_csv(r'C:\Users\User\Desktop\credit-risk-project\data\processed\train_cleaned.csv', parse_dates=['disbursement_date', 'due_date'])
```

```python
train.head()
```

```python
# FINANCIAL FEATURES i.e interest, ratio of repayment n lender profit
train['interest_amount'] = train['Total_Amount_to_Repay'] - train['Total_Amount']
train['repayment_ratio'] = train['Total_Amount_to_Repay'] / train['Total_Amount']
train['lender_profit'] = train['Lender_portion_to_be_repaid'] - train['Amount_Funded_By_Lender']
```

```python
# DATE FEATURES i.e loan month, loan day of the week n loar quater of the yr
train['loan_month'] = train['disbursement_date'].dt.month
train['loan_dayofweek'] = train['disbursement_date'].dt.dayofweek
train['loan_quarter'] = train['disbursement_date'].dt.quarter
```

```python
# LOAN FREQUENCY - loan counts throughout the dataset
loan_counts = train.groupby('customer_id')['tbl_loan_id'].count().reset_index()
loan_counts.columns = ['customer_id', 'customer_loan_count']
train = train.merge(loan_counts, on='customer_id', how='left')
```

```python
# DROP REDUNDANT/REPETITIVE COLUMNS
drop_cols = ['ID', 'country_id', 'Total_Amount_to_Repay', 'Lender_portion_to_be_repaid', 'disbursement_date', 'due_date']
train = train.drop(columns=drop_cols)
print(train.shape)
print(train.columns.tolist())
```

```python
# ENCODING CATEGORICALS - DUMMIES - HOTENCODER
# Categorical text data into numericals - transform loan type n new vs repeat. Dropping first category for each column,
# 2 possible type to give back one column 0 n 1 to rep
train = pd.get_dummies(train, columns=['loan_type', 'New_versus_Repeat'], drop_first=True)
print(train.shape) # number of added features
print(train.dtypes) # verify 
```

```python
# SAVE FEATURES
train.to_csv(r'C:\Users\User\Desktop\credit-risk-project\data\processed\train_features.csv', index=False)
print("Saved successfully")
```
